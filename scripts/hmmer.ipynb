```python
%%bash

wget -nc http://ftp.ebi.ac.uk/pub/databases/Pfam/releases/Pfam35.0/Pfam-A.hmm.gz

gunzip -f Pfam-A.hmm.gz

hmmpress Pfam-A.hmm

hmmfetch --index Pfam-A.hmm
```

```python
%%bash
ls -lh Pfam-A.hmm*
grep -m 3 "PF00145" Pfam-A.hmm
```

```python
pfams = [
    "PF00145",  # DNMT
    "PF01429",  # MBD
    "PF00850",  # HDAC
    "PF00856",  # SET
    "PF06325",  # PRMT
    "PF02182",  # UHRF
    "PF00628",  # PHD
    "PF00439",  # Bromodomain
    "PF00176",  # SNF2
    "PF00271",  # Helicase
    "PF00385",  # Chromodomain
]

full_pfams = []

with open("Pfam-A.hmm") as f:
    for line in f:
        if line.startswith("ACC"):
            acc = line.split()[1]
            if acc.split(".")[0] in pfams:
                full_pfams.append(acc)

full_pfams = sorted(set(full_pfams))

with open("target_pfams_full.txt", "w") as f:
    for acc in full_pfams:
        f.write(acc + "\n")

print("\n".join(full_pfams))
```

```python
%%bash
rm -f epigenetic_domains.hmm

while read pfam
do
    echo "Fetching $pfam"
    hmmfetch Pfam-A.hmm "$pfam" >> epigenetic_domains.hmm
done < target_pfams_full.txt

ls -lh epigenetic_domains.hmm
hmmpress -f epigenetic_domains.hmm
hmmstat epigenetic_domains.hmm
```

```python
import pandas as pd

rows = []

with open("hmmer_epigenetic_domtblout.txt") as f:
    for line in f:
        if line.startswith("#"):
            continue

        parts = line.split()

        protein = parts[0]
        domain = parts[3]
        evalue = float(parts[6])

        description = " ".join(parts[22:])

        rows.append(
            {
                "protein": protein,
                "domain": domain,
                "evalue": evalue,
                "description": description,
            }
        )

df = pd.DataFrame(rows)

df = (
    df.sort_values("evalue")
      .drop_duplicates(subset=["protein", "domain"])
)

print(df.shape)
df.head(20)
```

```python
selected_genes = [
    "DNA (cytosine-5)-methyltransferase 1",
    "DNA (cytosine-5)-methyltransferase 3B",
    "UHRF1",
    "methyl-CpG-binding domain protein 2",
    "histone deacetylase 4",
    "histone deacetylase 2",
    "histone deacetylase 3",
    "histone deacetylase 6",
    "SETDB1",
    "EHMT2",
    "EZH2",
    "SUV39H2",
    "protein arginine N-methyltransferase 1",
    "protein arginine N-methyltransferase 6",
    "protein polybromo-1",
    "chromodomain-helicase-DNA-binding protein 1",
    "SWI/SNF-related matrix-associated actin-dependent regulator"
]
```

```python
feature_table = pd.read_csv(
    "/content/GCF_006345805.1_ASM634580v1_feature_table.txt",
    sep="\t",
    low_memory=False
)

feature_table = feature_table.fillna("")

matches = []

for gene in selected_genes:
    mask = (
        feature_table["name"].str.contains(
            gene,
            case=False,
            regex=False
        )
    )

    found = feature_table[mask]

    if len(found):
        matches.append(
            found[
                [
                    "name",
                    "symbol",
                    "chromosome",
                    "genomic_accession",
                    "start",
                    "end",
                    "strand",
                    "GeneID"
                ]
            ]
        )

result = pd.concat(matches)

result.drop_duplicates().head(100)
```

```python
print(result.shape)
result.head(20)
```

```python
result["gene_length"] = result["end"] - result["start"]

result_unique = (
    result
    .sort_values("gene_length", ascending=False)
    .drop_duplicates(subset=["GeneID"])
)

print(result_unique.shape)

result_unique[
    [
        "name",
        "symbol",
        "chromosome",
        "start",
        "end",
        "gene_length"
    ]
]
```

```python
print("Всего строк:", len(result))
print("Уникальных GeneID:", result["GeneID"].nunique())
```

```python
final_table = result_unique[
    [
        "name",
        "symbol",
        "chromosome",
        "genomic_accession",
        "start",
        "end",
        "strand",
        "GeneID",
    ]
].copy()

final_table = final_table.rename(columns={
    "name": "Gene name",
    "symbol": "Gene symbol",
    "chromosome": "Chromosome",
    "genomic_accession": "Accession",
    "start": "Start",
    "end": "End",
    "strand": "Strand",
})

final_table.to_csv("epigenetic_genes_final.tsv", sep="\t", index=False)

final_table
```
